**Perform the following operations using Python on the data sets**

**a. Data cleaning**

**b. Data integration**

**c. Data transformation**

**d. Error correcting**

**e. Data model building**

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("your_dataset.csv")


# ----------------------------------------
# a. Data Cleaning
# ----------------------------------------

# Rename columns
df = df.rename(columns={'old_name': 'new_name'})

# Drop unwanted columns
df = df.drop(['col_to_drop'], axis=1)

# Remove duplicates
df = df.drop_duplicates()

# Handle missing values
df = df.dropna()   # or use df.fillna(method='ffill')

# Convert data types (example)
df['col1'] = df['col1'].astype(float)


# ----------------------------------------
# b. Data Integration
# ----------------------------------------

# Load another dataset
df2 = pd.read_csv("dataset2.csv")

# Merge datasets (common column required)
merged_df = pd.merge(df, df2, on='common_column', how='inner')

# OR concatenate
concat_df = pd.concat([df, df2])


# ----------------------------------------
# c. Data Transformation
# ----------------------------------------

# Scaling / transformation
df['col1'] = df['col1'] * 100

# Log transformation
df['col2'] = np.log(df['col2'])

# Create new feature
df['new_col'] = df['col1'] + df['col2']


# ----------------------------------------
# d. Error Correcting
# ----------------------------------------

# Replace incorrect values
df['col1'] = df['col1'].replace(-999, np.nan)

# Fix inconsistent formats
df['col2'] = df['col2'].replace(',', '.', regex=True)

# Remove outliers (IQR method)
Q1 = df['col1'].quantile(0.25)
Q3 = df['col1'].quantile(0.75)
IQR = Q3 - Q1
df = df[(df['col1'] >= Q1 - 1.5*IQR) & (df['col1'] <= Q3 + 1.5*IQR)]


# ----------------------------------------
# e. Data Model Building
# ----------------------------------------

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Define features and target
X = df.drop('target_column', axis=1)
y = df['target_column']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

#Replace 'col1', 'target_column', 'common_column' with actual column names.
```
